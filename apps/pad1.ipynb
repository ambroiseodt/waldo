```python
import mteb

benchmark_name = "Multilingual"
task_type = "Classification"
max_n_docs = 8192

# Get the full English benchmark
benchmark = mteb.get_benchmark(f"MTEB({benchmark_name})")

# Filter to only retrieval tasks
tasks = mteb.filter_tasks(benchmark, task_types=[task_type])
# Filter out large tasks if max_n_documents is set
if max_n_docs is not None:
    reduced_tasks = []
    for task in tasks:
        print(task.metadata.name)
        train_n_docs = (
            task.metadata.descriptive_stats["train"]["num_samples"] if "train" in task.metadata.descriptive_stats else 0
        )
        test_n_docs = (
            task.metadata.descriptive_stats["test"]["num_samples"] if "test" in task.metadata.descriptive_stats else 0
        )
        print(train_n_docs, test_n_docs)
        if max(train_n_docs, test_n_docs) <= max_n_docs:
            reduced_tasks.append(task)

tasks = reduced_tasks
print(len(tasks), "tasks after filtering")
domains = set()
languages = set()
for task in tasks:
    domains.update(task.metadata.domains)
    languages.update(task.metadata.languages)
print("domains:", len(domains), domains)
print("languages:", len(languages), languages)
```
